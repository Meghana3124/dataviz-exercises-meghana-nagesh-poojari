## Task 1 - CO2 per Capita for All Countries

All countries are shown in grey, with the United States highlighted in red.


```python

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

df = pd.read_csv('co2_emissions.csv')

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Countries:", df['Country'].nunique())
print("Years:", df['Year'].min(), "-", df['Year'].max())

df.head()
```

```python

highlight_country = 'United States'

fig = go.Figure()

# All other countries in grey
for country in df['Country'].unique():
    if country != highlight_country:
        country_data = df[df['Country'] == country]

        fig.add_trace(go.Scatter(
            x=country_data['Year'],
            y=country_data['CO2_per_capita'],
            mode='lines',
            line=dict(color='#EEEEEE', width=1),
            showlegend=False
        ))

# Highlight United States
highlight_data = df[df['Country'] == highlight_country]

fig.add_trace(go.Scatter(
    x=highlight_data['Year'],
    y=highlight_data['CO2_per_capita'],
    mode='lines',
    line=dict(color='#D62728', width=3),
    showlegend=False
))

# Label at the end of the line
last_point = highlight_data.iloc[-1]

fig.add_annotation(
    x=last_point['Year'],
    y=last_point['CO2_per_capita'],
    text=highlight_country,
    showarrow=False,
    xshift=8,
    font=dict(color='#D62728', size=12),
    xanchor='left'
)

fig.update_layout(
    title="United States: CO₂ per capita has declined over time",
    xaxis_title="Year",
    yaxis_title="CO₂ per capita",
    template="plotly_white"
)

fig.update_xaxes(showgrid=False, showline=False)
fig.update_yaxes(showgrid=False, showline=False)

fig.show()
```

## Task 2 - CO2 Emissions Over Time for Asian Countries

All Asian countries are shown in grey, with India highlighted in red.

```python

# Task 2: CO2 emissions for Asian countries

# Filter Asian countries
asia_df = df[df['Region'] == 'Asia'].copy()

# Highlight India
highlight_country = 'India'

fig = go.Figure()

# All other Asian countries in grey
for country in asia_df['Country'].unique():

    if country != highlight_country:
        country_data = asia_df[asia_df['Country'] == country]

        fig.add_trace(go.Scatter(
            x=country_data['Year'],
            y=country_data['CO2_Mt'],
            mode='lines',
            line=dict(color='#EEEEEE', width=1),
            showlegend=False
        ))

# Highlight India in red
highlight_data = asia_df[
    asia_df['Country'] == highlight_country
]

fig.add_trace(go.Scatter(
    x=highlight_data['Year'],
    y=highlight_data['CO2_Mt'],
    mode='lines',
    line=dict(color='#D62728', width=3),
    showlegend=False
))

# Direct label at the end of the line
last_point = highlight_data.iloc[-1]

fig.add_annotation(
    x=last_point['Year'],
    y=last_point['CO2_Mt'],
    text=highlight_country,
    showarrow=False,
    xshift=8,
    font=dict(color='#D62728', size=12),
    xanchor='left'
)

# Chart styling
fig.update_layout(
    title="India: CO₂ emissions have increased substantially",
    xaxis_title="Year",
    yaxis_title="CO₂ emissions (Mt)",
    template="plotly_white"
)

# Remove gridlines and borders
fig.update_xaxes(showgrid=False, showline=False)
fig.update_yaxes(showgrid=False, showline=False)

fig.show()
```
